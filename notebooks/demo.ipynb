# Context-aware speech understanding

Speak a household scene. Several people may talk at once about different
things. For every phrase it heard, the robot decides whether that phrase is the
**instruction**, **context** that changes what the instruction means, or
**noise** to ignore — and then carries out what is left.

## First run

1. Leave **Robot** on `hsrb` and **Environment** on `apartment`.
2. Click **Start world** and wait for `World ready`. About 100 seconds; this
   also loads the speech and planning models.
3. Click **Play sample scene**.

Everything after that runs on its own: speech regions, voices, roles, the
counted instruction, the plan, and its execution in RViz. Give it a few
minutes — the language model runs on CPU.

The shipped recording holds three phrases by two speakers. Two forks are asked
for, one is already on the table, so the planner is told
`Put two forks from the drawer on the table. Bring exactly 1 Fork.`

## The controls

**Robot** and **Environment** — pick before starting the world; both lock while
a world stands. `hsrb` in the `apartment` is the tested combination.

**Start world** — builds the world, publishes it to RViz, loads the models.
Wait for `World ready` before anything else.

**Stop world** — shuts it down. Do this between runs: a previous run leaves
objects where it put them, and the next instruction is grounded against a world
that has already changed.

Four ways to give it a scene, all taking the same route afterwards:

| control | what it does |
|---|---|
| **Record (stops on silence)** | measures the room, then records and ends by itself 1.5 s after you stop talking |
| ⏺ **and the small player** | manual recording: click to start, click again to stop |
| **Audio clip** | upload a `.wav`, `.webm`, `.mp3` or `.ogg`; the reproducible route for evaluation |
| **Play sample scene** | the recording shipped in this repository, for demos without a microphone |

**Understand scene** and **Plan and execute** are for going step by step
instead. The first runs only the speech front-end and moves nothing; the second
takes its sentence, asks for a plan and hands it to the world.

## Reading the output

**Voice activity** — the waveform with the kept regions shaded. Gaps were
dropped before transcription, because Whisper invents words over silence.

**Heard phrases** — one row per phrase: its number, when it was said, which
voice (`?` when too short to judge), the transcription, the role
(`Instruction`, `Context`, `Ignored`, or `—` when nothing was judged), and for
context phrases how the task changes.

**Voice distances** — how far apart the voices measured. `0` is the same
direction, `1` unrelated. Green is below the cutoff and judged one voice, red
above it. This is the evidence behind the speaker numbers: a verdict says "one
voice", a distance of 0.22 against a cutoff of 0.50 also says the decision was
comfortable.

**Counted changes** — what each context phrase asked to change, summed per
object type, and the resulting number.

**Instruction to planner** — the single sentence the planner receives. This is
where the speech work ends; everything after it is the inherited planning and
execution half.

## If something goes wrong

**Nothing happens after recording.** The first run loads models and takes
minutes; the status line says what it is doing.

**Grounding errors when executing.** A previous run moved things. **Stop
world**, then **Start world**.

**Your own voice comes out as two speakers.** The distance exceeded the cutoff
— short exclamations next to spoken sentences are what does it. The numbers are
in the voice distance table.

**RViz is not visible.** Switch off `Simple` at the bottom left; the `Desktop`
tab appears next to this notebook.
