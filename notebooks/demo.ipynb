# Context-aware speech understanding

Speak a household scene, or upload a recording, and let the pipeline decide
per utterance whether it is the instruction, relevant context, or noise.

1. Pick a **robot** and an **environment**, then **Start world**. Wait for
   "World ready" before continuing. The choice is locked while the world
   stands; **Stop world** releases it so you can build a different one.
2. **Record (stops on silence)** listens, ends by itself once the talking
   stops, and then runs the whole chain on its own: speech regions, speakers,
   roles, the counted instruction, the plan, and its execution.
3. **Play sample scene** does the same with a recording shipped in this
   repository, for showing the demo where no microphone is available.
4. To go step by step instead, upload a clip and use **Understand scene** and
   **Plan and execute**.

limits: see `README.md`.
